Loads the analysis JSON produced by notebook 4b and produces:
- Per-run period characterisation tables
- Per-run anomaly tables
- One figure per period: time-series with anomaly markers + persistence diagram panels

## 0. Imports

```python
import sys, os
sys.path.insert(0, os.path.join(os.path.dirname(''), 'functions'))

import warnings
warnings.filterwarnings('ignore')

from period_visualisation_functions import (
    load_analysis,
    visualise_all,
)

print('Imports OK ✓')
```

## 1. Configuration  ← **edit this cell**

```python
VIZ_INPUT_FILE = 'analysis_contractBreakdown_Simple_V1_2020_2025.json'

# Which anomaly methods to overlay
# Options: 'sesd', 'iqr', 'zscore', 'rolling_sigma', 'iforest', 'consensus'
VIZ_METHODS = ['sesd']

# Which homology dimensions to show in persistence diagrams
# [1] = H1 only (loops);  [0, 1] = H0 + H1
VIZ_PD_DIMS = [1]

# How many 'normal' days to show PDs for per period
VIZ_NORMAL_DAYS = 3

# Figure width in inches
VIZ_FIG_WIDTH = 18

print(f'Input   : {VIZ_INPUT_FILE}')
print(f'Methods : {VIZ_METHODS}')
print(f'PD dims : {VIZ_PD_DIMS}')
print('Configuration set ✓')
```

## 2. Load analysis

```python
analysis = load_analysis(VIZ_INPUT_FILE)
print(f'Loaded  : {VIZ_INPUT_FILE}')
print(f'Runs    : {list(analysis["runs"].keys())}')
print(f'CP dates: {analysis["change_points"]}')
```

## 3. Visualise all runs

```python
visualise_all(
    analysis        = analysis,
    viz_methods     = VIZ_METHODS,
    viz_pd_dims     = VIZ_PD_DIMS,
    viz_normal_days = VIZ_NORMAL_DAYS,
    viz_fig_width   = VIZ_FIG_WIDTH,
)
```
